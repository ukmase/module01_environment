##Мини-эксперимент: влияние порядка выполнения кодовых ячеек на выполнение всей программы

```python
project_title = "Mini Corpus"
```

Ячейка кода выше присваивает переменной project_title значение "Mini Corpus".

```python
print(project_title)
```

Ячейка кода выше выводит значение переменной project_title, если она определена;
иначе выдаёт name_error.

##Что я проверил

В ходе работы я проверил, как порядок выполнения ячеек в файлах notebook влияет на порядок выполнения команд программы и её работоспособность, и убедился, что при неправильном порядке выполнения ячеек программа может работать неправильно или вовсе не работать.


##Таблица состояний

###При правильном порядке выполнения:
Шаг | Введённая команда | Изменения runtime | Вывод


---

1     | project_title = "Mini Corpus" | создано имя project_title со строковым значением | отсутствует


---


2 | print(project_title) | отсутствуют | Mini Corpus

###При неправильном порядке выполнения:
Шаг | Введённая команда | Изменения runtime | Вывод


---

1 | print(project_title) | отсутствуют | NameError: name 'project_title' is not defined



---

2     | project_title = "Mini Corpus" | создано имя project_title со строковым значением | отсутствует


##Как проверить

Если ячейка print(project_title) выводит Mini Corpus, то ячейки были запущены в правильном порядке; если NameError: name 'project_title' is not defined - в неправильном.

##Обычный тест и ошибка


Вывод обычного теста (сначала ячейка с project_title = "Mini Corpus", потом - с print(project_title)): Mini Corpus



Вывод ошибки (любой другой порядок выполнения ячеек): NameError: name 'project_title' is not defined

##Fixed cells

Ячейка кода выше присваивает переменной project_title значение "Text Lab", переводит его в верхний регистр и выводит результат.

##Работа с файлами

```python
mini_corpus = open('mini_corpus.tsv', encoding="cp1251") # Открываем файл
lines = []
for line in mini_corpus:
  lines.append(line.strip()) # Добавляем все строки в список
print(lines[0])
print(lines[1])
```

Ячейка кода выше открывает файл mini_corpus.tsv, добавляет все его строки в список lines и выводит первые две строки.
